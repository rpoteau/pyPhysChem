## [B] Python relies on APIs!<a name="API"></a>

### [B.1] What is an API?

<div class="introE">

<table width="100%" cellpadding="6"><tr><td bgcolor="#e6e6e6"><span style="color:#000000">📖 <b>Introduction</b></span></td></tr></table>

An **API** (*Application Programming Interface*) is a set of rules and functions that allows two programs to communicate.  
👉 Analogy: it's like the **menu at a restaurant**  
- You (the user) choose a dish → **the request**  
- The kitchen (the cook) prepares it → **the processing**  
- The waiter (the API) brings you the dish → **the response**  

You don't need to know how the kitchen works or to be creative in cooking, only how to read the menu (the API documentation). You may even be assisted by a generative AI to read that documentation, and to apply more easily the modules and tools the API provides.

</figure>
</div>

### [B.2] The power of APIs

<div class="introE">

<table width="100%" cellpadding="6"><tr><td bgcolor="#e6e6e6"><span style="color:#000000">📖 <b>Introduction</b></span></td></tr></table>

Let's return to the culinary analogy. You don't need to know how the kitchen works or to be creative in cooking, only how to read the menu (the API documentation).  
  
Thanks to this, you can "invite your friends" to enjoy a great meal without being a Michelin-starred chef: in the digital world, this means tapping into powerful and reliable resources without being an expert in the underlying code.  
 
**APIs democratize access to scientific complexity by making it simple to adopt and use.**
</div>


### [B.3] APIs within Python

<div class="introE">

<table width="100%" cellpadding="6"><tr><td bgcolor="#e6e6e6"><span style="color:#000000">📖 <b>Introduction</b></span></td></tr></table>

Python relies on APIs at every level:
- **Library APIs**:  
  Each library (NumPy, Pandas, RDKit…) exposes functions and classes that make up its API.  
```python
  import math
  math.sqrt(16)   # math library API
```

- **System APIs**:  
  Reading/writing files, accessing the operating system, working with networks.
- **External APIs**:  
  Python can query external services (e.g. Materials Project, PubChem) through their web APIs.

In short: Python is built on APIs — you use them constantly without needing to know the underlying code
</div>

### [B.4] Why APIs matter (for a student or a researcher)

<div class="introE">

<table width="100%" cellpadding="6"><tr><td bgcolor="#e6e6e6"><span style="color:#000000">📖 <b>Introduction</b></span></td></tr></table>

- **Don't reinvent the wheel**  
  → Instead of rewriting an algorithm or recompiling a database, you directly use the functions made available by the API.
- **Save time and reduce errors**  
  → APIs provide **tested and reliable** functions.  
  Example: `numpy.linalg.eig` to compute eigenvalues, rather than coding a decomposition yourself.
- **Access complex resources**  
  → APIs give access to enormous databases (*Materials Project*, *PubChem*…) or to heavy-duty systems (servers, clusters) that you couldn't handle directly.
- **Standardize exchanges**  
  → An API defines a **clear and documented** way to interact with a service. Everyone speaks the same language.
- **Interoperability**  
  → Thanks to APIs, Python can communicate with programs written in C, Fortran, Java, or with remote web services, without worrying about the internal details.
</div>
